```python
def gender_features(word):
    return {'last_letter': word[-1] if word else ''}

gender_features('obama')
```

```python
import nltk
nltk.download('names', quiet=True)

from nltk.corpus import names

all_names = names.words()
print(len(all_names))

labeled_names = (
    [(name, 'male') for name in names.words('male.txt')] +
    [(name, 'female') for name in names.words('female.txt')]
)
```

```python
from nltk.corpus import names

names_list = names.words()
print(names_list[:5])
```

```python
print(labeled_names[:5])
```

```python
import random

random.shuffle(labeled_names)

featuresets = [
    (gender_features(name), gender)
    for name, gender in labeled_names
]

train_set = featuresets[2000:]
test_set = featuresets[:2000]
```

```python
import nltk

classifier = nltk.NaiveBayesClassifier.train(train_set)

print(classifier.classify(gender_features('Chaitanya')))
print(nltk.classify.accuracy(classifier, test_set))
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vect = CountVectorizer(binary=True)
corpus = [
    "Tesseract is a good optical character recognition engine",
    "Optical character recognition is significant"
]

vect.fit(corpus)

vocab = vect.vocabulary_

for key in sorted(vocab.keys()):
    print("{}:{}".format(key, vocab[key]))

print(vect.transform(["This is a good optical illusion"]).toarray())

from sklearn.metrics.pairwise import cosine_similarity

document_1 = vect.transform([
    "Google Cloud Vision is a character recognition engine"
])
document_2 = vect.transform([
    "OCR is an optical character recognition engine"
])

similarity = cosine_similarity(document_1, document_2)
print(similarity)
```

```python
from urllib import request
import ssl
import certifi

url = "https://www.gutenberg.org/"

context = ssl.create_default_context(cafile=certifi.where())

req = request.Request(
    url,
    headers={'User-Agent': 'Mozilla/5.0'}
)

response = request.urlopen(req, context=context, timeout=20)
raw = response.read().decode('utf-8', errors='replace')

print(type(raw))
print(len(raw))
print(raw[:75])
```
